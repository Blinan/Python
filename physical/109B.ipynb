```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 设置支持中文的字体（Windows 常用 SimHei，macOS 常用 PingFang SC，Linux 常用 WenQuanYi Micro Hei）
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'PingFang SC', 'WenQuanYi Micro Hei']
# 解决负号显示为方块的问题
plt.rcParams['axes.unicode_minus'] = False
```

```python
theta = np.linspace(0,75,6)
I = np.array([1.875,1.470,1.375,0.6918,0.420,0.100])

fig, (ax1,ax2,ax3) = plt.subplots(1,3,figsize=(15,7),layout='constrained')

ax1.set_title('验光光强I随检偏器角度θ的变化曲线')
ax1.set_xlabel(r'$\theta/ _{}^{0}$')
ax1.set_ylabel(r'$I/uW$')
ax1.plot(theta, I)

ax2.set_title('光强I随角度θ余弦值平方的线性图')
ax2.set_xlabel(r'$cos^{2}(\theta)$')
ax2.set_ylabel(r'I/uW')
ax2.plot(np.cos(np.deg2rad(theta)), I)

ax3.set_xlabel(r'ln(cos(\theta))')
ax3.set_ylabel(r'lnI')
ax3.plot(np.log(np.cos(theta)),np.log(I))

plt.show()
```
