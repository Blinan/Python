```python
# The programme add table of book(also make it clean )
# coding: utf-8
import os
from pathlib import Path
import shutil
import re

ReExp = re.compile(r'.*\.\D+')

sourcePath = Path('D:\\ssss\\1')
counter=1
for folderName, subfolders, filenames in os.walk(sourcePath):
    for filename in filenames:
        counter+=1
        filePath = Path.cwd()/filename
        filterFilename = ReExp.search(filePath.name)
        newFilename = (Path.cwd()/(filterFilename.group())).stem+str(counter)+(Path.cwd()/filterFilename.group()).suffix
        shutil.move(filePath,Path.cwd()/(newFilename))
        print(filename)
        pass
```

```python
from pypdf import PdfWriter
import re
from pathlib import Path


def addBooknotes():
    writer = PdfWriter(clone_from="py.pdf")
    
    fileObj = open(Path('booknotes.txt'), 'r', encoding='utf-8')
    fileLines = fileObj.readlines()
    
    parentReExp = re.compile(r'第\S+\s*\S*')
    ReExp = re.compile(r'(\d\.)+\d\s+\S+')
    
    pageReExp = re.compile('\d+(\s)?$')
    
    for fileLine in fileLines:
        
        parSearch = parentReExp.search(fileLine)
        Search = ReExp.search(fileLine)
        pageSearch = pageReExp.search(fileLine)
        print(f'parSearch = {parSearch},Search = {Search}, pageSearch = {pageSearch}')
        if not parSearch == None and not pageSearch == None:
            introduce = writer.add_outline_item(
                title=parSearch.group(),
                page_number=int(pageSearch.group())+13
            )
        elif not Search == None and not pageSearch == None:
            writer.add_outline_item(
                title=Search.group(),
                page_number=int(pageSearch.group())+13,
                parent=introduce
            )
    
    writer.write("colonPy.pdf")
    
    print("DONE!")
    pass
addBooknotes()
```
